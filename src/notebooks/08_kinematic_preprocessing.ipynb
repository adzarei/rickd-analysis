# Kinematic Preprocessing

## Preprocessing Steps

This notebook focuses on enriching the dataset with kinematic and kinetic features derived from marker data. The main steps include:

 - Calculating joint and segment angles using inverse kinematics.
 - Identifying gait cycle events (swing and stance phases) by detecting heel-strike (HS) and toe-off (TO) events with a velocity-based algorithm.
 - Segmenting the data into stance and swing phases, and considering whether to analyze full cycles or focus on specific phases (potentially guided by a Silency Map).
 - Normalizing the timeseries of gait cycle phases so that all cycles have the same number of samples (e.g., via cubic interpolation).
 - Estimating kinetic data using inverse dynamics.
 - For each swing, stance, or whole segment, calculating descriptive variables (as reported in relevant literature) and peak angular velocities (as derivatives of the angles).
 - Adding participant metadata such as age, dominant leg, gender, height, weight, and history of running-related injuries.
 - Computing descriptive variables from the perspective of the dominant leg and quantifying differences between the dominant and non-dominant legs.
