```python
RESULTS_DIR = Path(c.RICKD_PROCESSED_DATA_FOLDER) / 'timeseries'
RESULTS_DIR.mkdir(exist_ok=True, parents=True)

session_ids = loader.get_successful_sessions()
print(f"Total successful sessions: {len(session_ids)}")

joints =  ['L_ankle', 'L_foot', 'L_hip', 'L_knee', 'R_ankle', 'R_foot', 'R_hip', 'R_knee', 'pelvis']

# Build channel ordering: per joint → angle XYZ then velocity XYZ
channels = []
for j in joints:
    for comp in ['X','Y','Z']:
        channels.append(f"{j}_angle_{comp}")
    for comp in ['X','Y','Z']:
        channels.append(f"{j}_velocity_{comp}")
```

```python
n_jobs = min(8, mp.cpu_count())
results_m = []
with ProcessPoolExecutor(max_workers=n_jobs) as ex:
    # Using the mean curve aggregator.
    futures = [ex.submit(compute_session_representation, sid, joints, MeanCurveAggregator()) for sid in session_ids]
    for f in tqdm(as_completed(futures), total=len(futures), desc=f"Processing sessions (process pool) - Num. Jobs: {n_jobs}"):
        r = f.result()
        if r is not None:
            results_m.append(r)

valid_session_ids_m, reps_m = zip(*results_m) if results_m else ([], [])
all_sessions_matrix_m = np.stack(reps_m, axis=0) if reps_m else np.zeros((0,101,0))
print(f"Matrix shape: {all_sessions_matrix_m.shape} (n_sessions, 101, {len(channels)})")

all_sessions_matrix_m
```

```python
# Persist results (Mean variant)
np.save(RESULTS_DIR / 'timeseries_mean_matrix.npy', all_sessions_matrix_m)
with open(RESULTS_DIR / 'timeseries_mean_channels.json', 'w') as f:
    json.dump(channels, f, indent=2)
with open(RESULTS_DIR / 'timeseries_mean_sessions.json', 'w') as f:
    json.dump(list(valid_session_ids_m), f, indent=2)
```

```python
n_jobs = min(8, mp.cpu_count())
results_cr = []
with ProcessPoolExecutor(max_workers=n_jobs) as ex:
    # Using the mean curve aggregator.
    aggregator = RegistrationCurveAggregator()
    futures = [ex.submit(compute_session_representation, sid, joints, aggregator) for sid in session_ids]
    for f in tqdm(as_completed(futures), total=len(futures), desc=f"Processing sessions (process pool) - Num. Jobs: {n_jobs}"):
        r = f.result()
        if r is not None:
            results_cr.append(r)

valid_session_ids_cr, reps_cr = zip(*results_cr) if results_cr else ([], [])
all_sessions_matrix_cr = np.stack(reps_cr, axis=0) if reps_cr else np.zeros((0,101,0))
print(f"Matrix shape: {all_sessions_matrix_cr.shape} (n_sessions, 101, {len(channels)})")

all_sessions_matrix_cr
```

```python
# Persist results (Curve Registration variant)
np.save(RESULTS_DIR / 'timeseries_cr_matrix.npy', all_sessions_matrix_cr)
with open(RESULTS_DIR / 'timeseries_cr_channels.json', 'w') as f:
    json.dump(channels, f, indent=2)
with open(RESULTS_DIR / 'timeseries_cr_sessions.json', 'w') as f:
    json.dump(list(valid_session_ids_cr), f, indent=2)
```
