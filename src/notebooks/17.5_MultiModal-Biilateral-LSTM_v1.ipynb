```python
def train_test_group_split(
    X_ts, X_meta, y, subject_id, test_size=0.2, random_state=None
):
    """
    Split data into train and test sets or train and validation sets using GroupShuffleSplit to ensure subject-level separation.
    Returns:
        X_ts_train, X_ts_test, X_meta_train, X_meta_test, y_train, y_test, subject_train, subject_test
    """
    splitter = GroupShuffleSplit(n_splits=1, test_size=test_size, random_state=random_state)
    train_idx, test_idx = next(splitter.split(X_ts, y, groups=subject_id))

    # Split the data
    X_ts_train, X_ts_test = X_ts[train_idx], X_ts[test_idx]
    X_meta_train, X_meta_test = X_meta[train_idx], X_meta[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
    subject_train, subject_test = subject_id[train_idx], subject_id[test_idx]

    print(f"Training set: {len(train_idx)} sessions from {len(np.unique(subject_train))} subjects")
    print(f"Test/Validation set: {len(test_idx)} sessions from {len(np.unique(subject_test))} subjects")

    # Verify no subject overlap
    subject_overlap = set(subject_train) & set(subject_test)
    if len(subject_overlap) > 0:
        raise ValueError("Subject overlap found in train-test split!!!")

    # Check class distribution in splits
    train_counts = np.bincount(y_train.astype(int))
    test_counts = np.bincount(y_test.astype(int))
    print(f"\nTrain class distribution: {train_counts} ({np.round(train_counts/len(y_train)*100, 2)}%)")
    print(f"Test/Validation class distribution: {test_counts} ({np.round(test_counts/len(y_test)*100, 2)}%)")

    return (
        X_ts_train, X_ts_test,
        X_meta_train, X_meta_test,
        y_train, y_test,
        subject_train, subject_test
    )

print("Train-Test Split")
print("="*50)
(
    X_ts_train, X_ts_test,
    X_meta_train, X_meta_test,
    y_train, y_test,
    subject_train, subject_test
) = train_test_group_split(X_ts_scaled, X_meta_scaled, y, subject_id, test_size=0.2, random_state=RANDOM_STATE)
```

## Building the model

```python
def build_model(hp: kt.HyperParameters,
                model_name: str,
                features: int = 25,
                time_steps: int = 101,
                metadata_features: int = 8,
                clipnorm: float = 1.0):
    """
    Returns a compiled Keras model. Inputs (by name):
      - "left":  (None, time_steps, features)
      - "right": (None, time_steps, features)
      - "metadata": (None, metadata_features)  # z-scored numeric vector
    """

    # ----------------- Tunables -----------------
    use_multiscale = hp.Boolean("multiscale_conv", default=True)
    with hp.conditional_scope("multiscale_conv", [False]):
        single_kernel = hp.Choice("single_kernel", [7, 15, 21, 31])

    lstm_units     = hp.Choice("lstm_units", [48, 64, 96])
    lstm_dropout   = hp.Float("lstm_dropout", 0.15, 0.35, step=0.05)

    W              = hp.Choice("num_windows", [4, 5, 6])  # stance windows

    win_lstm_units = hp.Choice("win_lstm_units", [16, 32, 48])
    win_lstm_drop  = hp.Float("win_lstm_dropout", 0.10, 0.30, step=0.05)

    head_hidden    = hp.Choice("head_hidden", [48, 64, 96])
    head_dropout   = hp.Float("head_dropout", 0.20, 0.40, step=0.05)

    lr             = hp.Float("lr", 1e-4, 5e-4, sampling="log")
    wd             = hp.Float("wd", 1e-5, 3e-4, sampling="log")

    # Small prefix to keep layer names unique if you build multiple models in one process
    pref = model_name if model_name else "model"

    # ----------------- Sub-models (shared) -----------------
    def build_unilateral_timestep_encoder():
        inp = L.Input((time_steps, features), name=f"{pref}_enc_in")
        if use_multiscale:
            x1 = L.Conv1D(8, 3,  padding="same", activation="relu", name=f"{pref}_enc_conv_k3")(inp)
            x2 = L.Conv1D(4, 9,  padding="same", activation="relu", name=f"{pref}_enc_conv_k9")(inp)
            x3 = L.Conv1D(4, 21, padding="same", activation="relu", name=f"{pref}_enc_conv_k21")(inp)
            x  = L.Concatenate(name=f"{pref}_enc_conv_concat")([x1, x2, x3])
        else:
            x  = L.Conv1D(16, single_kernel, padding="same", activation="relu", name=f"{pref}_enc_conv_single")(inp)
        x = L.Dropout(0.10, name=f"{pref}_enc_conv_dropout")(x)
        x = L.Bidirectional(
                L.LSTM(lstm_units, return_sequences=True, dropout=lstm_dropout),
                name=f"{pref}_enc_bilstm_t"
            )(x)  # (B,T, 2*lstm_units)
        return M.Model(inp, x, name=f"{pref}_unilateral_timestep_encoder")

    def build_window_summarizer():
        """(B,T,2U) -> (B,W,4U) via per-window avg⊕max pooling."""
        tseq = L.Input((time_steps, 2*lstm_units), name=f"{pref}_ws_in")
        base, rem = time_steps // W, time_steps % W
        outs, start = [], 0
        for i in range(W):
            sz = base + (1 if i < rem else 0)
            s, e = start, start + sz
            win = L.Lambda(lambda t, s=s, e=e: t[:, s:e, :], name=f"{pref}_ws_slice_{i}")(tseq)
            avg = L.GlobalAveragePooling1D(name=f"{pref}_ws_avg_{i}")(win)
            mx  = L.GlobalMaxPooling1D(name=f"{pref}_ws_max_{i}")(win)
            v   = L.Concatenate(name=f"{pref}_ws_vec_{i}")([avg, mx])            # (B,4U)
            v   = L.Lambda(lambda z: tf.expand_dims(z, 1), name=f"{pref}_ws_expand_{i}")(v)  # (B,1,4U)
            outs.append(v); start = e
        out = L.Concatenate(axis=1, name=f"{pref}_windows_concat")(outs)         # (B,W,4U)
        return M.Model(tseq, out, name=f"{pref}_window_summarizer")

    # 1) window head now knows the feature dim (d_win) and allows variable #windows
    def build_window_head(d_win: int, win_lstm_units: int, win_lstm_drop: float, prefix: str):
        """(B, W, d_win) -> (B, 64), with variable W but fixed feature width d_win."""
        inp = L.Input((None, d_win), name=f"{prefix}_wh_in")  # time=None, features=d_win
        x = L.Bidirectional(
                L.LSTM(win_lstm_units, return_sequences=True, dropout=win_lstm_drop),
                name=f"{prefix}_wh_bilstm"
            )(inp)
        a = L.Dense(1, activation="tanh", name=f"{prefix}_wh_attn_score")(x)
        a = L.Softmax(axis=1, name=f"{prefix}_wh_attn_softmax")(a)
        attn = L.Lambda(lambda z: tf.reduce_sum(z[0]*z[1], axis=1),
                        name=f"{prefix}_wh_attn_pool")([x, a])      # (B, 2*win_lstm_units)
        mx = L.GlobalMaxPooling1D(name=f"{prefix}_wh_gmp")(x)        # (B, 2*win_lstm_units)
        h  = L.Concatenate(name=f"{prefix}_wh_feat")([attn, mx])     # (B, 4*win_lstm_units)
        out = L.Dense(64, activation="relu", name=f"{prefix}_wh_emb")(h)
        return M.Model(inp, out, name=f"{prefix}_window_head")

    # 2) MAG fuser keeps a fixed W because RepeatVector needs it
    def build_mag_fuser(W: int, d_win: int, prefix: str):
        """Fuse (B, W, d_win) with (B, 32) via a gate."""
        H_win = L.Input((W, d_win), name=f"{prefix}_mag_in_win")
        h_tab = L.Input((32,),      name=f"{prefix}_mag_in_tab")
        tab_proj = L.Dense(d_win, activation="relu", name=f"{prefix}_mag_tab_proj")(h_tab)
        tab_rep  = L.RepeatVector(W, name=f"{prefix}_mag_tab_repeat")(tab_proj)   # (B, W, d_win)
        z = L.Concatenate(name=f"{prefix}_mag_concat")([H_win, tab_rep])          # (B, W, 2*d_win)
        g = L.Dense(d_win, activation="sigmoid", name=f"{prefix}_mag_gate")(z)    # (B, W, d_win)
        fused = L.Add(name=f"{prefix}_mag_fused")([
            L.Multiply()([g, H_win]),
            L.Multiply()([L.Lambda(lambda x: 1.0 - x)(g), tab_rep])
        ])
        return M.Model([H_win, h_tab], fused, name=f"{prefix}_mag_fuser")


    # Instantiate shared parts
    d_win = 4 * lstm_units  # because each window vector is [avg ⊕ max] of (2*lstm_units)

    # shared sub-models
    enc   = build_unilateral_timestep_encoder()
    ws    = build_window_summarizer()
    mag   = build_mag_fuser(W, d_win, prefix=pref)
    whead = build_window_head(d_win, win_lstm_units, win_lstm_drop, prefix=pref)

    # ----------------- Inputs -----------------
    L_in   = L.Input((time_steps, features), name="left")
    R_in   = L.Input((time_steps, features), name="right")
    meta_in = L.Input((metadata_features,),  name="metadata")

    # ----------------- Graph -----------------
    # Per-timestep encoders
    xL = enc(L_in)                # (B,T,2U)
    xR = enc(R_in)

    # Window summarization to (B,W,4U)
    HL = ws(xL)
    HR = ws(xR)

    # Tabular compact embedding (→ 32)
    h_tab = L.Dense(64, activation="relu", name=f"{pref}_tab_dense1")(meta_in)
    h_tab = L.Dropout(0.10,             name=f"{pref}_tab_dropout")(h_tab)
    h_tab = L.Dense(32, activation="relu", name=f"{pref}_tab_dense2")(h_tab)

    # MAG fusion per side (shared fuser)
    fL = mag([HL, h_tab])         # (B,W,4U)
    fR = mag([HR, h_tab])

    # Window temporal heads (shared head)
    hL = whead(fL)                # (B,64)
    hR = whead(fR)

    # Symmetry-aware merge
    diff  = L.Subtract(name=f"{pref}_diff")([hL, hR])
    adiff = L.Lambda(tf.abs, name=f"{pref}_abs_diff")(diff)
    prod  = L.Multiply(name=f"{pref}_prod")([hL, hR])
    merged = L.Concatenate(name=f"{pref}_merge")([hL, hR, adiff, prod])  # (B,256)

    # Head
    x = L.Dense(head_hidden, activation="relu", name=f"{pref}_head_dense")(merged)
    x = L.Dropout(head_dropout, name=f"{pref}_head_dropout")(x)
    out = L.Dense(1, activation="sigmoid", name="p_injured")(x)

    model = M.Model({"left": L_in, "right": R_in, "metadata": meta_in}, out, name=model_name or "bilat_mag")

    # Optimizer (AdamW preferred; fallback to Adam)
    try:
        opt = tf.keras.optimizers.AdamW(
            learning_rate=lr, weight_decay=wd,
            global_clipnorm=clipnorm  # prefer global
        )
    except TypeError:
        opt = tf.keras.optimizers.Adam(learning_rate=lr, clipnorm=clipnorm)

    model.compile(
        optimizer=opt,
        loss="binary_crossentropy",
        metrics=[
            tf.keras.metrics.AUC(curve="PR",  name="auc_pr"),
            tf.keras.metrics.AUC(curve="ROC", name="auc_roc"),
            tf.keras.metrics.BinaryAccuracy(name="accuracy"),
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall")
        ],
    )
    return model
```
